# Custom Tool

```python
from langchain_core.tools import tool
```

```python
@tool
def multiply(a: int, b: int) -> int :
    "Multiply two numbers"
    return a*b
```

```python
result = multiply.invoke({"a": 8, "b": 7})
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
```

# Structured Tools

```python
from langchain.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The First Number to be multiplied")
    b: int = Field(required=True, description="The Second Number to be multiplied")
```

```python
def multiply_func(a:int, b:int) -> int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name="multiply",
    description="Multiply two Numbers",
    args_schema=MultiplyInput
)
```

```python
result = multiply_tool.invoke({"a": 3, "b": 4})
print(result)
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# Using BaseTool Class

```python
from langchain_core.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(required=True, description="The first number to add")
    b: int = Field(required=True, description="The second number to add")
```

```python
class MultiplyTool(BaseTool):
    name: str = "multiply"
    description: str = "Multiply two numbers"
    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int):
        return a*b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({'a': 3, 'b': 4})
print(result)
```

```python
print(multiply_tool.name)
print(multiply_tool.description)
print(multiply_tool.args)
```

# ToolKit

```python
from langchain_core.tools import tool

@tool
def add(a: int, b: int):
    "Add two numbers."
    return a + b

@tool
def multiply(a: int, b: int):
    "Multiply two numbers."
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, '=>', tool.description)
```
